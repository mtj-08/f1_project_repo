```sql
With driver_metrics AS(
 SELECT driver_name, sum(race_starts) AS race_starts, sum(number_of_wins) AS total_wins, SUM(number_of_podiums) AS total_podiums, SUM(CASE WHEN standings =1 THEN 1 ELSE 0 END) AS total_championships from v_driver_standings group by driver_name having total_championships>=1
)
select driver_name, race_starts, total_wins, total_podiums, total_championships, (total_championships*100) + (total_wins *10) + (total_podiums *3 ) AS greatness_score from driver_metrics order by greatness_score DESC;
```
