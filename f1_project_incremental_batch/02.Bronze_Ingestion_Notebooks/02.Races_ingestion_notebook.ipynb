```python
source_file = f"{landing_folder_path}/{v_batch_id}/races.csv"
table_name = f"{catalog_name}.{bronze_schema}.races"
```

```python
from pyspark.sql.types import StructField, StringType, IntegerType,DateType, StructType
races_schema = StructType([
    StructField('season', IntegerType()),
    StructField('round', IntegerType()),
    StructField('url', StringType()),
    StructField('raceName', StringType()),
    StructField('date', DateType()),
    StructField('circuitId', StringType())
])
```

### 2. Reading the races.csv file from the files location

```python
races_df = (
    spark.read
    .format('csv')
    .option('header','true')
    .schema(races_schema)
    .load(source_file)
     )
     
```

### 3. View the circuit df

```python
display(races_df)
```

### 4. Adding the file ingestion timestamp and sourcefile metadata in the df

```python
final_races = add_timestamp_metadata(races_df)
```

### 5. Writing the final dataframe into the table under bronze schema

```python
write_to_bronze (
input_df=final_races,
target_table = table_name,
batch_id = v_batch_id
)
```
