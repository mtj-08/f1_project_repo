```python
source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

```python
con_schema = """constructorId STRING, name STRING, nationality STRING, url STRING"""
```

### 2. Reading the constructors.json file from the files location

```python
cons_df = (
    spark.read
    .format('json')
    .option('mode','FAILFAST')
    .schema(con_schema)
    .load(source_file)
     )
```

### 3. View the cons df

```python
display(cons_df)
```

### 4. Adding the file ingestion timestamp and sourcefile metadata in the df

```python
final_cons = add_timestamp_metadata(cons_df)
```

### 5. Writing the final dataframe into the table under bronze schema

```python
(
    final_cons.write.format('delta')
    .mode('overwrite')
    .saveAsTable(table_name)
)
```
